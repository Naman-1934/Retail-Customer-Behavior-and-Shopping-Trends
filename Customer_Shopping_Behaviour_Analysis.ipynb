```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the dataset

```python
df = pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

## Data Pre-Processing

```python
df.info()
```

```python
# Summary Statistics

df.describe()

# If you want categorical column's statistical measures
# df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median))
```

```python
df.isnull().sum()
```

#### Converting the column names into more remember format


```python
# First, lowering the column names

df.columns = df.columns.str.lower()

# Now, replace all spaces with the underscore ( _ )
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
# purchase_amount_(usd) looks weird and different. So, rename it

df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

## Feature Engineering

```python
# Create a new column age_group

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.cut(df['age'], labels=labels, bins=4)


# Another function for the same
# df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df.head()
```

```python
df['frequency_of_purchases'].unique()
```

```python
# Create a new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Every 3 Months:': 90,
    'Annually': 365
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
# Converting from float to integer and use fillna because there will be some NA (Not Available) values

df['purchase_frequency_days'] = df['purchase_frequency_days'].fillna(0).astype(int)
```
